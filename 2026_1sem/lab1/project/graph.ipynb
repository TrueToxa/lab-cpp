```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def plot_distr(df, name="", save=0):
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    ax.bar(df["N"], df["p"])
    ax.set_xlabel("N")
    ax.set_ylabel("p")
    ax.set_title(f"Propability distribution of dice = {name}")
    if save:
        plt.savefig("fig/" + name + ".png", dpi=300)
```

```python
df = pd.read_csv("csv/penalty100.csv")
plot_distr(df)
# df
```

```python
import os
import glob

for path in glob.glob("csv/*.csv"):
    name = os.path.splitext(os.path.basename(path))[0]

    df = pd.read_csv(path)

    plot_distr(df, name, 1)
```
